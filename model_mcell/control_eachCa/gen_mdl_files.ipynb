```python
import numpy as np
import os

dirLoc = '/home/nishant/lab/MFB/model_mcell/control_eachCa/'
setupLoc = '/home/nishant/lab/MFB/data/findPr/'
template = os.path.join(dirLoc, 'xmain.mdl')

with open(template, 'r') as file:
    fc = file.read()

print(fc)
```

```python
def getSimInfo(dname, key=None, skip=1):
    infoList = dname.split('_')[skip:]
    if key:
        info = infoList[infoList.index(key)+1]
    else:
        info = dict([[infoList[i], infoList[i+1]] for i in range(0,len(infoList),2)])
    return info

control_dirs = os.listdir('/home/nishant/lab/MFB/data/control')

attr = ['nVDCC', 'dVAZ', 'nAZ']
control_set = []
for control_dir in control_dirs[:]:
    s = getSimInfo(control_dir, skip=0)
    temp = {}
    for a in attr:
        temp.update({a: s[a]})

    if temp not in control_set:
        control_set.append(temp)


# print(len(control_set), '\n', control_set)

# setup = np.genfromtxt(os.path.join(setupLoc, 'controls.dat'), usecols=(1,2,3,4), dtype='int')
#print(setup)


#############################################333
saveFile = True
ISI = [20]
i=0
nap = 10

for isi in ISI:
    for control_dir in control_set:
        
        nvdcc = control_dir['nVDCC']
        dvaz = control_dir['dVAZ']
        naz   = control_dir['nAZ']
        # rrp   = s['RRP']

        simTime = isi*nap + 10
        #print(simTime)
        replace_text = [['nVDCC = "5"', f'nVDCC = "{nvdcc}"'],
                        ['dVDCC = "100"',f'dVDCC = "{dvaz}"'],
                        ['nAZ = "18"',f'nAZ = "{naz}"'],
                        ['ITERATIONS = 150e3', f'ITERATIONS = {simTime}e3'],
                        ['nAP = "6"', f'nAP = "{nap}"'],
                        ['isi = "20"', f'isi = "{isi}"']]

        newText = fc
        for rt in replace_text:
            newText = newText.replace(rt[0], rt[1])

        f = f"xmain_nVDCC_{nvdcc}_dVAZ_{dvaz}_nAZ_{naz}_ISI_{isi}_nAP_{nap}.mdl"
        i+=1
        print(i,f)
        
        # print(newText)
        if(saveFile): 
            fname = os.path.join(dirLoc, f)
            with open(fname, 'w+') as file:
                file.write(newText)
```
