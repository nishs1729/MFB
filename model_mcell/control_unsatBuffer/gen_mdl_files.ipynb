```python
from itertools import product
import numpy as np
import os, sys

sys.path.append('../../scripts_analysis')
from misc import *

dirLoc = '../../model_mcell/control_unsatBuffer/'
setupLoc = '../../model_mcell'
template_file_loc = os.path.join(dirLoc, 'xmain.mdl')

with open(template_file_loc, 'r') as file:
    template_text = file.read()

print(template_text)
```

```python
def gen_mdl_file(fname, replace_text, new_text=template_text, save=False):
    for rt in replace_text:
        new_text = new_text.replace(rt[0], rt[1])

    if save:
        f_loc = os.path.join(dirLoc, fname)
        with open(f_loc, 'w+') as file:
            file.write(new_text)

    return new_text
```

```python
designs = np.genfromtxt(os.path.join(setupLoc, 'representative_designs.txt'), dtype=str)
# print(designs)

for d in designs[:]:
    info = getSimInfo(d, skip=0)
    print(d)
    naz   = int(info['nAZ'])
    dvaz  = int(info['dVAZ'])
    rrp   = int(info['RRP'])
    nvdcc = int(info['nVDCC'])
    isi   = int(info['ISI'])
    nap   = int(info['nAP'])

    simTime = isi*nap + 10

    fname = f"xmain_nVDCC_{nvdcc}_dVAZ_{dvaz}_nAZ_{naz}_RRP_{rrp}_ISI_{isi}_nAP_{nap}.mdl"

    replace_text = [['nVDCC = "5"', f'nVDCC = "{nvdcc}"'],
                    ['dVAZ = "100"',f'dVAZ = "{dvaz}"'],
                    ['nAZ = "18"',f'nAZ = "{naz}"'],
                    ['RRP = "20"',f'RRP = "{rrp}"'],
                    ['ITERATIONS = 150e3', f'ITERATIONS = {simTime}e3'],
                    ['nAP = "10"', f'nAP = "{nap}"'],
                    ['isi = "20"', f'isi = "{isi}"']]
    
    print(gen_mdl_file(fname, replace_text, save=True))
```
