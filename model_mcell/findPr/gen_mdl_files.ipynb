```python
from itertools import product
import os

cwd = os.getcwd()

dirLoc = cwd
template = os.path.join(dirLoc, 'xmain.mdl')

with open(template, 'r') as file:
    fc = file.read()

print(fc)
```

```python
nvdcc = range(1, 16)
dvdcc = range(60, 231, 20)
naz   = range(7, 36, 2)
rrp   = range(5, 61, 5)

for nVDCC, dVDCC, nAZ in product(nvdcc, dvdcc, naz):
    replace_text = [['nVDCC = "1"', f'nVDCC = "{nVDCC}"'],
                    ['dVDCC = "60"',f'dVDCC = "{dVDCC}"'],
                    ['nAZ = "7"',f'nAZ = "{nAZ}"']]
                   # ['RRP = "7"', f'RRP = "{RRP}"']
    newText = fc
    for rt in replace_text:
        newText = newText.replace(rt[0], rt[1])
        
    f = f"xmain_nVDCC_{nVDCC}_dVDCC_{dVDCC}_nAZ_{nAZ}.mdl"
    print(f)
    fname = os.path.join(dirLoc, f)
    with open(fname, 'w+') as file: file.write(newText)
```
