```python
from itertools import product
import numpy as np
import os

dirLoc = '/home/nishant/lab/MFB/mcell/control/'
setupLoc = '/home/nishant/lab/MFB/results/findPr/'
template = os.path.join(dirLoc, 'xmain.mdl')

with open(template, 'r') as file:
    fc = file.read()

print(fc)
```

```python
setup = np.genfromtxt(os.path.join(setupLoc, 'controls_all.dat'), usecols=(0,1,2,3), dtype='int')

saveFile = True
i=0
for s in setup:
    naz   = s[2]
    dvdcc = s[1]
    rrp   = s[3]
    nvdcc = s[0]

    nap = 10
    ISI = [20]
    
    for isi in ISI:
        simTime = isi*nap + 10
        #print(simTime)
        replace_text = [['nVDCC = "5"', f'nVDCC = "{nvdcc}"'],
                        ['dVDCC = "100"',f'dVDCC = "{dvdcc}"'],
                        ['nAZ = "18"',f'nAZ = "{naz}"'],
                        ['RRP = "20"',f'RRP = "{rrp}"'],
                        ['ITERATIONS = 150e3', f'ITERATIONS = {simTime}e3'],
                        ['nAP = "6"', f'nAP = "{nap}"'],
                        ['isi = "20"', f'isi = "{isi}"']]

        newText = fc
        for rt in replace_text:
            newText = newText.replace(rt[0], rt[1])

        f = f"xmain_nVDCC_{nvdcc}_dVDCC_{dvdcc}_nAZ_{naz}_RRP_{rrp}_ISI_{isi}_nAP_{nap}.mdl"
        i+=1
        print(i,f)
        
        if(saveFile): 
            fname = os.path.join(dirLoc, f)
            with open(fname, 'w+') as file:
                file.write(newText)
```
